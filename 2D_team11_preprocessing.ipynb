OUTLIERS HANDLING

->we are calculating lower limit  and upper limit..

->below lower limit and above upper limit is considerd as outlier..

->by using  IQR method(INTER QUARTILE RANGE).



```python
Q1=df['Monthly Charge'].quantile(0.25)
Q3=df['Monthly Charge'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
outliers=df[(df["Monthly Charge"]<lower)|(df['Monthly Charge']>upper)]
print(len(outliers))
```

```python
Q1=df['Total Charges'].quantile(0.25)
Q3=df['Total Charges'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
outliers=df[(df["Total Charges"]<lower)|(df['Total Charges']>upper)]
print(len(outliers))
```
